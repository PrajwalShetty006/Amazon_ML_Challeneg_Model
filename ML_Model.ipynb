```python
import re
import unicodedata

def normalize_text(text):
    if pd.isna(text):
        return ""
    
    text = str(text).lower()
    
    # Normalize unicode
    text = unicodedata.normalize("NFKD", text)
    
    # Remove accents
    text = "".join(
        c for c in text 
        if not unicodedata.combining(c)
    )
    
    # & -> and
    text = text.replace("&", " and ")
    
    # Keep only letters and numbers
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    
    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()
    
    return text
```

```python
def find_required_records(
    path,
    required_ids,
    already_found=None,
    chunksize=100_000
):
    if already_found is None:
        already_found = set()

    remaining = required_ids - already_found
    results = []

    print("Need to find:", len(remaining))

    for chunk_no, chunk in enumerate(
        pd.read_csv(
            path,
            sep="\t",
            chunksize=chunksize
        )
    ):
        matched = chunk[
            chunk["entity_id"].isin(remaining)
        ]

        if not matched.empty:
            results.append(matched)

            found_now = set(matched["entity_id"])
            remaining -= found_now

        print(
            f"Chunk {chunk_no + 1:02d} | "
            f"Found: {len(required_ids) - len(remaining):,}/"
            f"{len(required_ids):,} | "
            f"Remaining: {len(remaining):,}"
        )

        if not remaining:
            break

    if results:
        return pd.concat(results, ignore_index=True)

    return pd.DataFrame()
```

```python
def extract_numbers(text):
    if pd.isna(text):
        return []
    
    return re.findall(r'\d+', str(text))
```

```python
for df in [source1_dev, source2_pool, source3_pool]:
    df["address_numbers"] = df["business_address"].apply(extract_numbers)
```

```python
def get_name_tokens(text):
    if not text:
        return []
    
    tokens = str(text).split()
    
    # Ignore very short tokens
    return [
        t for t in tokens
        if len(t) >= 3
    ]
```

```python
name_token_index_s2 = defaultdict(list)
name_token_index_s3 = defaultdict(list)

address_number_index_s2 = defaultdict(list)
address_number_index_s3 = defaultdict(list)
```

```python
for idx, row in source2_pool.iterrows():
    
    # Name tokens
    for token in set(get_name_tokens(row["name_norm"])):
        name_token_index_s2[token].append(idx)
    
    # Address numbers
    for number in set(row["address_numbers"]):
        address_number_index_s2[number].append(idx)
```

```python
for idx, row in source3_pool.iterrows():
    
    for token in set(get_name_tokens(row["name_norm"])):
        name_token_index_s3[token].append(idx)
    
    for number in set(row["address_numbers"]):
        address_number_index_s3[number].append(idx)
```

```python
def get_candidates(row, max_candidates=500):
    
    candidates_s2 = set()
    candidates_s3 = set()
    
    # -------------------------
    # NAME TOKEN BLOCKING
    # -------------------------
    
    name_tokens = get_name_tokens(row["name_norm"])
    
    for token in name_tokens:
        
        candidates_s2.update(
            name_token_index_s2.get(token, [])
        )
        
        candidates_s3.update(
            name_token_index_s3.get(token, [])
        )
    
    
    # -------------------------
    # ADDRESS NUMBER BLOCKING
    # -------------------------
    
    for number in row["address_numbers"]:
        
        candidates_s2.update(
            address_number_index_s2.get(number, [])
        )
        
        candidates_s3.update(
            address_number_index_s3.get(number, [])
        )
    
    
    # -------------------------
    # COUNTRY FILTER
    # -------------------------
    
    candidates_s2 = {
        idx for idx in candidates_s2
        if source2_pool.loc[idx, "country_norm"]
        == row["country_norm"]
    }
    
    candidates_s3 = {
        idx for idx in candidates_s3
        if source3_pool.loc[idx, "country_norm"]
        == row["country_norm"]
    }
    
    
    # Safety cap
    if len(candidates_s2) > max_candidates:
        candidates_s2 = set(list(candidates_s2)[:max_candidates])
    
    if len(candidates_s3) > max_candidates:
        candidates_s3 = set(list(candidates_s3)[:max_candidates])
    
    return candidates_s2, candidates_s3
```
